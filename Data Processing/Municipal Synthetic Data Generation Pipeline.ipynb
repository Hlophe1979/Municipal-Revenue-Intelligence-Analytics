We'll generate 100,000 billing records covering 24 months, with realistic variation in municipalities, services, billing amounts, payments, arrears, customer types, and payment behaviour.

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
municipalities = [
    ("Gauteng", "Municipality_A"),
    ("Gauteng", "Municipality_B"),
    ("Limpopo", "Municipality_C"),
    ("Limpopo", "Municipality_D"),
    ("Mpumalanga", "Municipality_E"),
    ("Mpumalanga", "Municipality_F"),
    ("KwaZulu-Natal", "Municipality_G"),
    ("KwaZulu-Natal", "Municipality_H"),
]

municipality_df = spark.createDataFrame(
    municipalities,
    ["province", "municipality"]
)

display(municipality_df)
```

```python
num_customers = 25000

customers = (
    spark.range(1, num_customers + 1)
    .withColumnRenamed("id", "customer_number")
    .withColumn(
        "customer_id",
        F.concat(F.lit("CUST"), F.lpad(F.col("customer_number"), 6, "0"))
    )
    .withColumn(
        "customer_type",
        F.when(F.rand(seed=10) < 0.82, "Residential")
         .otherwise("Business")
    )
    .withColumn(
        "property_type",
        F.when(F.rand(seed=20) < 0.65, "House")
         .when(F.rand(seed=30) < 0.85, "Apartment")
         .otherwise("Commercial")
    )
    .withColumn(
        "province",
        F.expr("""
            CASE
                WHEN rand() < 0.35 THEN 'Gauteng'
                WHEN rand() < 0.55 THEN 'Limpopo'
                WHEN rand() < 0.75 THEN 'Mpumalanga'
                ELSE 'KwaZulu-Natal'
            END
        """)
    )
)

display(customers.limit(20))
```

```python
customers = (
    customers
    .withColumn(
        "municipality",
        F.expr("""
            CASE
                WHEN province = 'Gauteng' AND rand() < 0.5
                    THEN 'Municipality_A'
                WHEN province = 'Gauteng'
                    THEN 'Municipality_B'

                WHEN province = 'Limpopo' AND rand() < 0.5
                    THEN 'Municipality_C'
                WHEN province = 'Limpopo'
                    THEN 'Municipality_D'

                WHEN province = 'Mpumalanga' AND rand() < 0.5
                    THEN 'Municipality_E'
                WHEN province = 'Mpumalanga'
                    THEN 'Municipality_F'

                WHEN province = 'KwaZulu-Natal' AND rand() < 0.5
                    THEN 'Municipality_G'
                ELSE 'Municipality_H'
            END
        """)
    )
)

display(customers.limit(20))
```

```python
months = (
    spark.range(0, 24)
    .withColumn(
        "billing_date",
        F.add_months(F.to_date(F.lit("2024-01-01")), F.col("id"))
    )
    .select("billing_date")
)

display(months)
```

```python
transactions = customers.crossJoin(months)

print("Number of records:", transactions.count())
```

```python
transactions = (
    transactions
    .withColumn(
        "service_type",
        F.when(F.rand(seed=101) < 0.40, "Electricity")
         .when(F.rand(seed=102) < 0.70, "Water")
         .when(F.rand(seed=103) < 0.90, "Waste")
         .otherwise("Property Rates")
    )
)

display(transactions.limit(20))
```

```python
transactions = (
    transactions
    .withColumn(
        "base_amount",
        F.when(
            F.col("service_type") == "Electricity",
            F.when(F.col("customer_type") == "Business", 3500).otherwise(1100)
        )
        .when(
            F.col("service_type") == "Water",
            F.when(F.col("customer_type") == "Business", 2200).otherwise(650)
        )
        .when(
            F.col("service_type") == "Waste",
            F.when(F.col("customer_type") == "Business", 1200).otherwise(300)
        )
        .otherwise(
            F.when(F.col("customer_type") == "Business", 2800).otherwise(900)
        )
    )
    .withColumn(
        "billing_amount",
        F.round(
            F.col("base_amount") *
            (F.lit(0.75) + F.rand(seed=200) * F.lit(0.75)),
            2
        )
    )
    .drop("base_amount")
)

display(transactions.limit(20))
```

```python
transactions = (
    transactions
    .withColumn(
        "payment_status",
        F.when(F.rand(seed=301) < 0.68, "Paid")
         .when(F.rand(seed=302) < 0.90, "Partial")
         .otherwise("Unpaid")
    )
)
```

```python
transactions = (
    transactions
    .withColumn(
        "payment_amount",
        F.when(
            F.col("payment_status") == "Paid",
            F.col("billing_amount")
        )
        .when(
            F.col("payment_status") == "Partial",
            F.round(
                F.col("billing_amount") *
                (F.lit(0.20) + F.rand(seed=401) * F.lit(0.60)),
                2
            )
        )
        .otherwise(F.lit(0.0))
    )
)
```

```python
transactions = (
    transactions
    .withColumn(
        "outstanding_balance",
        F.round(
            F.col("billing_amount") - F.col("payment_amount"),
            2
        )
    )
)
```

```python
transactions = (
    transactions
    .withColumn(
        "payment_method",
        F.when(F.col("payment_status") == "Unpaid", "None")
         .when(F.rand(seed=501) < 0.45, "EFT")
         .when(F.rand(seed=502) < 0.75, "Card")
         .when(F.rand(seed=503) < 0.90, "Cash")
         .otherwise("Debit Order")
    )
)
```

```python
transactions = (
    transactions
    .withColumn(
        "outstanding_balance",
        F.round(
            F.col("billing_amount") - F.col("payment_amount"),
            2
        )
    )
)
```

```python
transactions = (
    transactions
    .withColumn(
        "payment_method",
        F.when(F.col("payment_status") == "Unpaid", "None")
         .when(F.rand(seed=501) < 0.45, "EFT")
         .when(F.rand(seed=502) < 0.75, "Card")
         .when(F.rand(seed=503) < 0.90, "Cash")
         .otherwise("Debit Order")
    )
)
```

```python
transactions = (
    transactions
    .withColumn(
        "days_overdue",
        F.when(
            F.col("payment_status") == "Paid",
            F.lit(0)
        )
        .when(
            F.col("payment_status") == "Partial",
            F.round(F.rand(seed=601) * 90).cast("int")
        )
        .otherwise(
            F.round(F.lit(30) + F.rand(seed=602) * 300).cast("int")
        )
    )
)
```

```python
final_df = transactions.select(
    "customer_id",
    "province",
    "municipality",
    "customer_type",
    "property_type",
    "service_type",
    "billing_date",
    "billing_amount",
    "payment_amount",
    "payment_method",
    "payment_status",
    "outstanding_balance",
    "days_overdue"
)
```

```python
display(final_df.limit(20))
```

```python
final_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("municipal_revenue")
```

## Analytical Query

```python
%sql

SELECT
    province,
    SUM(billing_amount) AS total_billed,
    SUM(payment_amount) AS total_collected,
    SUM(outstanding_balance) AS total_outstanding,
    ROUND(
        SUM(payment_amount) / SUM(billing_amount) * 100,
        2
    ) AS collection_rate
FROM municipal_revenue
GROUP BY province
ORDER BY collection_rate DESC;
```
